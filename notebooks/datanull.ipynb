# Preparación del CSV para Cassandra

Este cuaderno prepara una copia del CSV para cargarla en Cassandra: sustituye valores ausentes por el texto `NULL` y exporta el índice como columna `Id`. Las rutas corresponden al entorno original y deben revisarse antes de ejecutarlo.

```python
import pandas as pd
```

```python
# Carga fichero
file = "/Users/carlos/Documents/proj_tokio/data/Air_Traffic_Passenger_Statistics.csv"
data = pd.read_csv(file, sep=',')
data
```

```python
# Sustituyo los valores ausentes por el texto que espera la carga de Cassandra.
data.fillna('NULL', inplace=True) # inplace modifica el dataframe original en vez de generar uno nuevo
```

```python
# Guardo el fichero en el contenedor de Docker
# Exporto el índice como Id para usarlo como clave primaria en Cassandra.
file_exit = "/var/lib/cassandra/Air_Traffic_Passenger_Statistics_null.csv"
data.to_csv(file_exit, index=True,index_label='Id', header=True)
```
